# Persian sentiment analysis

Examples use fixed responses. Set `LIVE = True` to call Mistral with your API key.

```python
from langchain_core.language_models.fake_chat_models import FakeListChatModel

from sentiment_analysis import SentimentAnalyzer

LIVE = False
model = None if LIVE else FakeListChatModel(
    responses=["Positive", '["Positive", "Negative", "Neutral"]', "True"]
)
analyzer = SentimentAnalyzer(chat_model=model)
```

```python
analyzer.classify("عالی بود")
```

```python
analyzer.classify(["عالی بود", "خیلی دیر رسید", "معمولی بود"], method="batch")
```

```python
analyzer.update_templates(
    single_template="Return True if the review concerns delivery, otherwise False: {sentence}",
    labels=("True", "False"),
)
analyzer.classify("بسته خیلی دیر رسید")
```
